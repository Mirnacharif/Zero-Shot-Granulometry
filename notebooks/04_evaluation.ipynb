# Segmentation Evaluation

Predicted masks are compared with their corresponding ground-truth annotations to compute Average Precision (AP) metrics using the evaluation functions provided by [ImageGrains](https://github.com/dmair1989/imagegrains).


```python
import sys
sys.path.insert(0, "..")

from pathlib import Path
from imagegrains import data_loader, segmentation_helper
import numpy as np

# Dataset configuration
PROJECT_ROOT = Path("..").resolve()
images_path = PROJECT_ROOT / "data" / "IG2" / "PP" / "test"

pred_str = "SAM_F0"

# Load images, ground-truth masks, and predictions
imgs, lbls, preds = data_loader.dataset_loader(
    images_path,
    label_str="mask",
    pred_str=pred_str
)

# Set evaluation filters
filters = {
    "edge": [False, 0.05],
    "px_cutoff": [True, 8]
}

# Evaluate segmentation performance
eval_results = segmentation_helper.eval_set(
    imgs,
    lbls,
    preds,
    data_id=pred_str,
    tar_dir=str(images_path),
    filters=filters,
    save_results=True,
    return_results=True
)
```

```python
# Print results for each image
for res in eval_results.values():
    print(f"{res['id']}: AP@0.5 = {res['ap'][0]:.2f}, "
          f"mAP = {np.mean(res['ap']):.2f}")

# Overall dataset results
ap05 = [res["ap"][0] for res in eval_results.values()]
maps = [np.mean(res["ap"]) for res in eval_results.values()]

print("\nDATASET RESULTS")
print(f"Number of images: {len(eval_results)}")
print(f"Mean AP@0.5: {np.nanmean(ap05):.2f}")
print(f"Mean mAP: {np.nanmean(maps):.2f}")
```
